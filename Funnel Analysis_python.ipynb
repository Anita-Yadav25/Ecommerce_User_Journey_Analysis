# Data Processing and Cleaning

```python
# Check for null and duplicate values

df.isnull().sum()
```

```python
# Finding duplicate values

df.duplicated().sum()
```

```python
# uniue data
df.nunique()
```

```python
# Fixing the data and extracting time base data

df['Date'] = df['Event Time'].dt.date
df['DayOfWeek'] = df['Event Time'].dt.day_name()
df['Hour'] = df['Event Time'].dt.hour
df['WeekNumber'] = df['Event Time'].dt.isocalendar().week
```

```python
df.head()
```

```python
df['Event_sequence'] = df.groupby('Session ID').cumcount()+ 1
```

```python
df.head(10)
```

```python

# Sort events by session and time
df = df.sort_values(
    ['Session ID', 'Event Time']
).reset_index(drop=True)

# Recalculate event sequence
df['Event_sequence'] = (
    df.groupby('Session ID').cumcount() + 1
)

print("Dataset shape:", df.shape)
print("Missing values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())
```

```python

df.to_csv("funnel_analysis_clean.csv", index=False)

print("Clean dataset exported successfully!")
```

```python

from sqlalchemy import create_engine, text

# Connect to the master database
engine_master = create_engine(
    f"mssql+pyodbc:///?odbc_connect={params}",
    isolation_level="AUTOCOMMIT"
)

# Create the database if it doesn't exist
with engine_master.connect() as connection:
    connection.execute(text("""
        IF DB_ID('FunnelAnalysisDB') IS NULL
            CREATE DATABASE FunnelAnalysisDB
    """))

print("FunnelAnalysisDB is ready!")
```

```python

# Connect to FunnelAnalysisDB
project_connection_string = connection_string.replace(
    "DATABASE=master;",
    "DATABASE=FunnelAnalysisDB;"
)

project_params = urllib.parse.quote_plus(
    project_connection_string
)

engine = create_engine(
    f"mssql+pyodbc:///?odbc_connect={project_params}"
)

with engine.connect() as connection:
    print("Connected to FunnelAnalysisDB successfully!")
```
